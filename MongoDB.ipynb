```python
import pandas as pd
import numpy as np
from pymongo import MongoClient
client = MongoClient("mongodb+srv://Wiritpon:User1234@cluster0.ummjbqm.mongodb.net/?appName=Cluster0")
```

```python
customers = pd.read_csv('customers_clean.csv')
orders = pd.read_csv('orders_clean.csv')
complaints = pd.read_csv('complaints_clean.csv')
app_events = pd.read_csv('app_events_clean.csv')

print("customers rows:", len(customers))
print("orders rows:", len(orders))
print("complaints rows:", len(complaints))
print("app_events rows:", len(app_events))
```

```python
#To get the customer's own profile row
customer_profile = customers[customers['customer_id'] == sample_customer_id]
print("Profile:")
print(customer_profile)

#To find this customer's orders
customer_orders = orders[orders['customer_id'] == sample_customer_id]
print("\nOrders count:", len(customer_orders))
print(customer_orders.head())

#To find this customer's complaints
customer_complaints = complaints[complaints['customer_id'] == sample_customer_id]
print("\nComplaints count:", len(customer_complaints))

#To find this customer's app events
customer_events = app_events[app_events['customer_id'] == sample_customer_id]
print("\nApp events count:", len(customer_events))
```

```python
#Get the one row of customer profile data
customer_profile = customers[customers['customer_id'] == sample_customer_id].iloc[0]

#Convert that row to a dictionary
profile_dict = customer_profile.to_dict()

print("Customer profile as a dictionary:")
print(profile_dict)
```

```python
#To get this customer's orders as a list of dictionaries
customer_orders = orders[orders['customer_id'] == sample_customer_id]
orders_list = customer_orders.to_dict(orient='records')

#To show what we got
print("Orders as a list of dictionaries:")
print(orders_list)

#To add it to our profile dictionary as a nested array
profile_dict['orders'] = orders_list
```

```python
customer_complaints = complaints[complaints['customer_id'] == sample_customer_id]
profile_dict['complaints'] = customer_complaints.to_dict(orient='records')

customer_events = app_events[app_events['customer_id'] == sample_customer_id]
profile_dict['app_events'] = customer_events.to_dict(orient='records')

print("Full customer document:")
print(profile_dict)
```

```python
all_customer_docs = []

customer_ids = customers['customer_id'].unique()

for cid in customer_ids:
    profile = customers[customers['customer_id'] == cid].iloc[0].to_dict()
    profile['orders'] = orders[orders['customer_id'] == cid].to_dict(orient='records')
    profile['complaints'] = complaints[complaints['customer_id'] == cid].to_dict(orient='records')
    profile['app_events'] = app_events[app_events['customer_id'] == cid].to_dict(orient='records')
    all_customer_docs.append(profile)

print("Total customer documents built:", len(all_customer_docs))
print("\nFirst document customer_id:", all_customer_docs[0]['customer_id'])
print("First document orders count:", len(all_customer_docs[0]['orders']))
```

```python
db = client['northstar']
customer_360 = db['customer_360']

customer_360.delete_many({})

result = customer_360.insert_many(all_customer_docs)

print("Documents inserted:", len(result.inserted_ids))
```

```python
incidents = pd.read_csv('incidents_clean.csv')

print("Total incidents:", len(incidents))
print("\nFirst incident:")
print(incidents.iloc[0])
```

```python
#Build a list of incident documents with embedded events_log arrays
all_incident_docs = []

#Create a loop through each incident row
for index, row in incidents.iterrows():
    #Start with the original incident fields as a dictionary
    doc = row.to_dict()

    #Build a synthesised events_log array for this incident
    events = []

#incident was reported
    events.append({
        "timestamp": row['reported_at'],
        "action": "reported",
        "actor": "system"
        })
#if resolved, add a resolved event
    if row['resolution_status'] == 'Resolved':
        events.append({
            "timestamp": row['reported_at'],
            "action": "resolved",
            "actor": "ops_team",
            "hours_to_resolve": row['resolved_hours']
        })
#Attach the events log to the document
    doc['events_log'] = events
    #Add the document to the list
    all_incident_docs.append(doc)

print("Total incident documents built:", len(all_incident_docs))
print("\nFirst incident document:")
print(all_incident_docs[0])
```

```python
#Get a reference to the incident_tracking collection
incident_tracking = db['incident_tracking']

#Clear any old data first
incident_tracking.delete_many({})

#Insert all incident documents
result = incident_tracking.insert_many(all_incident_docs)

print("Incident documents inserted:", len(result.inserted_ids))
```

```python
#To find all High or Critical severity incidents that aren't yet Resolved
unresolved_severe = list(incident_tracking.find({
    "severity": {"$in": ["High", "Critical"]},
    "resolution_status": {"$ne": "Resolved"}
}))

print("Unresolved high/critical incidents:", len(unresolved_severe))

#Show the first one
print("\nExample incident:", unresolved_severe[0]['incident_id'])
print("Severity:", unresolved_severe[0]['severity'])
print("Status:", unresolved_severe[0]['resolution_status'])
print("Events so far:", len(unresolved_severe[0]['events_log']))
```

```python
#Count incidents per delivery using a loop
all_incidents = list(incident_tracking.find({}))

incident_counts = {}
for doc in all_incidents:
    delivery_id = doc['delivery_id']
    if delivery_id in incident_counts:
        incident_counts[delivery_id] = incident_counts[delivery_id] + 1
    else:
        incident_counts[delivery_id] = 1

#Print every delivery
#We will seek for the highest ones manually
print("Incidents per delivery:")
for delivery_id in incident_counts:
    if incident_counts[delivery_id] >= 2:
        print(delivery_id, "had", incident_counts[delivery_id], "incidents")
```

```python
#Measure how long one query takes (before adding an index)
import time

start = time.time()
result = customer_360.find_one({"customer_id": "C0500"})
end = time.time()

query_time_ms = (end - start) * 1000

print("Query time without index:", query_time_ms, "ms")
```

```python
#Ask MongoDB how it's executing the query
plan = customer_360.find({"customer_id": "C0500"}).explain()

#Print the strategy MongoDB chose
winning_plan = plan['queryPlanner']['winningPlan']
print("Winning plan stage:")
print(winning_plan)
```

```python
#To create an index on customer_id to speed up customer lookups
index_name = customer_360.create_index("customer_id")

print("Index created:", index_name)

#List all indexes on the collection to verify
print("\nAll indexes on customer_360:")
for index in customer_360.list_indexes():
    print(" -", index['name'], index['key'])
```

```python
#Measure how long one query takes (after adding the index)
start = time.time()
result = customer_360.find_one({"customer_id": "C0500"})
end = time.time()

query_time_ms = (end - start) * 1000

print("Query time with index:", query_time_ms, "ms")
```

```python
#Check the execution plan again, now that the index exists
plan = customer_360.find({"customer_id": "C0500"}).explain()

winning_plan = plan['queryPlanner']['winningPlan']
print("Winning plan stage:")
print(winning_plan)
```
